## LETRA G

```
ganho_hora = float(input("Quanto você ganha por hora? R$ "))
horas_mes = float(input("Quantas horas trabalhou no mês? "))

salario_bruto = ganho_hora * horas_mes
ir = salario_bruto * 0.11
inss = salario_bruto * 0.08
sindicato = salario_bruto * 0.05
descontos = ir + inss + sindicato
salario_liquido = salario_bruto - descontos

print(f"Salário bruto: R$ {salario_bruto:.2f}")
print(f"Pagou ao INSS: R$ {inss:.2f}")
print(f"Pagou ao sindicato: R$ {sindicato:.2f}")
print(f"Salário líquido: R$ {salario_liquido:.2f}")
```

## LETRA H

```
import math

r = float(input("Digite o raio do tanque em metros: "))
h = float(input("Digite a altura do tanque em metros: "))

area_lateral = 2 * math.pi * r * h
area_bases = 2 * math.pi * r**2
area_total = area_lateral + area_bases

litros_tinta = area_total / 3
latas = math.ceil(litros_tinta / 5)
custo = latas * 50

print(f"Área total a pintar: {area_total:.2f} m²")
print(f"Quantidade de tinta necessária: {litros_tinta:.2f} litros")
print(f"Quantidade de latas: {latas}")
print(f"Custo total: R$ {custo:.2f}")
```

## LETRA I

```
import math

x1 = float(input("Digite x1: "))
y1 = float(input("Digite y1: "))
x2 = float(input("Digite x2: "))
y2 = float(input("Digite y2: "))

d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print(f"A distância entre os pontos é: {d:.2f}")
```

## LETRA J

```
import math

r = float(input("Digite o raio da caixa d'água em metros: "))
h = float(input("Digite a altura da caixa d'água em metros: "))

volume_m3 = math.pi * r**2 * h
volume_litros = volume_m3 * 1000
tempo_horas = volume_litros / 1350

print(f"O volume da caixa d'água é: {volume_litros:.2f} litros")
print(f"A autonomia é de: {tempo_horas:.2f} horas")
```

## LETRA K

```
arquivo_mb = float(input("Digite o tamanho do arquivo em MB: "))
velocidade_mbs = float(input("Digite a velocidade da internet em MB/s: "))

tempo_total_segundos = arquivo_mb / velocidade_mbs
minutos = int(tempo_total_segundos // 60)
segundos = int(tempo_total_segundos % 60)

print(f"Tempo aproximado de download: {minutos} minuto(s) e {segundos} segundo(s)")
```

```
a = float(input("Digite o primeiro valor: "))
b = float(input("Digite o segundo valor: "))
c = float(input("Digite o terceiro valor: "))

# Verificação de valores iguais
if a == b or a == c or b == c:
    print("Os valores devem ser distintos!")
else:
    numeros = [a, b, c]
    numeros.sort(reverse=True)  # do maior pro menor

    media = (numeros[0] + numeros[1]) / 2

    print(f"A média dos dois maiores valores é: {media:.2f}")
```
